```python
import pandas as pd
df = pd.read_csv('meta.csv')
df = df[df['dx'].isin(['mel', 'nv'])]
df = df.sort_values('image_id')
df = df.drop_duplicates('lesion_id')
df = df[(df['dx'] == 'mel') | ((df['dx'] == 'nv') & (df['dx_type'] == 'histo'))]

parcalar = []
for ds in df['dataset'].unique():
    kaynak = df[df['dataset'] == ds]
    mel = kaynak[kaynak['dx'] == 'mel']
    nv = kaynak[kaynak['dx'] == 'nv']
    nv_secilen = nv.sample(n=len(mel), random_state=42)
    print(ds, '-> mel:', len(mel), ' nv:', len(nv_secilen))
    parcalar.append(mel)
    parcalar.append(nv_secilen)

secim = pd.concat(parcalar).sort_values('image_id')
print('Toplam mel:', (secim['dx'] == 'mel').sum())
print('Toplam nv:', (secim['dx'] == 'nv').sum())
secim.to_csv('secim.csv', index=False)
```

```python
import pandas as pd, glob, os
secim = pd.read_csv('secim.csv')

img_yol = {os.path.splitext(os.path.basename(p))[0]: p
           for p in glob.glob('img/**/*.jpg', recursive=True)}
mask_yol = {os.path.basename(p).replace('_segmentation.png', ''): p
            for p in glob.glob('**/*_segmentation.png', recursive=True)}

print('Bulunan görüntü dosyası:', len(img_yol))
print('Bulunan maske dosyası:', len(mask_yol))

secim['img'] = secim['image_id'].map(img_yol)
secim['mask'] = secim['image_id'].map(mask_yol)
print('Seçimde satır:', len(secim))
print('Görüntüsü eksik:', secim['img'].isna().sum())
print('Maskesi eksik:', secim['mask'].isna().sum())
print(secim[['image_id', 'img', 'mask']].head(2).to_string())
```

```python
import cv2, numpy as np, matplotlib.pyplot as plt
from scipy import ndimage as ndi

def dice(a, b):
    kesisim = np.logical_and(a, b).sum()
    toplam = a.sum() + b.sum()
    return 2 * kesisim / toplam if toplam > 0 else 1.0

def hat(yol):
    bgr = cv2.imread(yol)
    gri = cv2.cvtColor(bgr, cv2.COLOR_BGR2GRAY)
    esik, _ = cv2.threshold(gri, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    ikili = gri < esik
    etiket, n = ndi.label(ikili)
    h, w = gri.shape
    if n == 0:
        return np.zeros((h, w), bool)
    secilen = etiket[h // 2, w // 2]
    if secilen == 0:
        merkezler = ndi.center_of_mass(ikili, etiket, range(1, n + 1))
        uzak = [(y - h / 2) ** 2 + (x - w / 2) ** 2 for y, x in merkezler]
        secilen = int(np.argmin(uzak)) + 1
    return ndi.binary_fill_holes(etiket == secilen)

sonuc, bos, uyumsuz = [], 0, 0
for _, s in secim.iterrows():
    uzman = cv2.imread(s['mask'], cv2.IMREAD_GRAYSCALE) > 127
    tahmin = hat(s['img'])
    if tahmin.shape != uzman.shape:
        uyumsuz += 1
        sonuc.append(np.nan)
        continue
    bos += tahmin.sum() == 0
    sonuc.append(dice(tahmin, uzman))
secim['dice_baz'] = sonuc

print('Boyut uyumsuz:', uyumsuz, ' Boş tahmin:', bos)
print(secim.groupby('dx')['dice_baz'].agg(['mean', 'std', 'count']))

fig, eks = plt.subplots(2, 3, figsize=(10, 6))
for i, dx in enumerate(['mel', 'nv']):
    s = secim[secim['dx'] == dx].iloc[0]
    eks[i, 0].imshow(cv2.cvtColor(cv2.imread(s['img']), cv2.COLOR_BGR2RGB))
    eks[i, 1].imshow(cv2.imread(s['mask'], cv2.IMREAD_GRAYSCALE) > 127, cmap='gray')
    eks[i, 2].imshow(hat(s['img']), cmap='gray')
    eks[i, 0].set_ylabel(dx)
    eks[i, 2].set_title('Dice = %.3f' % s['dice_baz'])
eks[0, 0].set_title('Görüntü')
eks[0, 1].set_title('Uzman maskesi')
plt.tight_layout()
plt.savefig('sekil1.png', dpi=150)
plt.show()
```

```python
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (7, 7))

def hat2(yol):
    m = hat(yol).astype(np.uint8)
    m = cv2.morphologyEx(m, cv2.MORPH_OPEN, kernel)
    m = cv2.morphologyEx(m, cv2.MORPH_CLOSE, kernel)
    return ndi.binary_fill_holes(m > 0)

sonuc2, bos2 = [], 0
for _, s in secim.iterrows():
    uzman = cv2.imread(s['mask'], cv2.IMREAD_GRAYSCALE) > 127
    tahmin = hat2(s['img'])
    bos2 += tahmin.sum() == 0
    sonuc2.append(dice(tahmin, uzman))
secim['dice_iyi'] = sonuc2
secim['fark'] = secim['dice_iyi'] - secim['dice_baz']

print('Boş tahmin:', bos2)
print(secim.groupby('dx')[['dice_baz', 'dice_iyi']].agg(['mean', 'std']))
print(secim.groupby('dx')['fark'].agg(artan=lambda x: (x > 0).sum(),
                                      azalan=lambda x: (x < 0).sum()))
```

```python
rows = []
for _, s in secim.iterrows():
    uzman = cv2.imread(s['mask'], cv2.IMREAD_GRAYSCALE) > 127
    t = hat(s['img'])
    kenar = t[0].any() or t[-1].any() or t[:, 0].any() or t[:, -1].any()
    rows.append((t.mean(), uzman.mean(), kenar))
secim['alan_t'] = [r[0] for r in rows]
secim['alan_u'] = [r[1] for r in rows]
secim['kenar'] = [r[2] for r in rows]
secim['basarisiz'] = secim['dice_baz'] < 0.05

print(secim.groupby(['dx', 'basarisiz']).size())
print(secim.groupby(['dx', 'basarisiz'])[['alan_t', 'alan_u']].median())
print(secim.groupby(['dx', 'basarisiz'])['kenar'].mean())
```

```python
rows = []
for _, s in secim[secim['basarisiz']].iterrows():
    gri = cv2.cvtColor(cv2.imread(s['img']), cv2.COLOR_BGR2GRAY)
    uzman = cv2.imread(s['mask'], cv2.IMREAD_GRAYSCALE) > 127
    esik, _ = cv2.threshold(gri, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    etiket, n = ndi.label(gri < esik)
    h, w = gri.shape
    en_buyuk = np.bincount(etiket.ravel())[1:].max() / (h * w) if n > 0 else 0
    rows.append((s['dx'], etiket[h // 2, w // 2] == 0, uzman[h // 2, w // 2],
                 n, en_buyuk, gri[h // 2, w // 2] - esik))
tani = pd.DataFrame(rows, columns=['dx', 'merkez_arka', 'merkez_uzmanda',
                                   'n_bolge', 'en_buyuk_oran', 'merkez_gri_fark'])
print(tani.groupby('dx')[['merkez_arka', 'merkez_uzmanda']].mean())
print(tani.groupby('dx')[['n_bolge', 'en_buyuk_oran', 'merkez_gri_fark']].median())
```

```python
def hat3(yol):
    gri = cv2.cvtColor(cv2.imread(yol), cv2.COLOR_BGR2GRAY)
    esik, _ = cv2.threshold(gri, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    etiket, n = ndi.label(gri < esik)
    h, w = gri.shape
    if n == 0:
        return np.zeros((h, w), bool), True
    secilen = etiket[h // 2, w // 2]
    yedek = secilen == 0
    if yedek:
        secilen = int(np.argmax(np.bincount(etiket.ravel())[1:])) + 1
    return ndi.binary_fill_holes(etiket == secilen), yedek

d3, yd = [], []
for _, s in secim.iterrows():
    uzman = cv2.imread(s['mask'], cv2.IMREAD_GRAYSCALE) > 127
    t, y = hat3(s['img'])
    d3.append(dice(t, uzman))
    yd.append(y)
secim['dice_h3'] = d3
secim['yedek'] = yd
secim['fark3'] = secim['dice_h3'] - secim['dice_baz']

print(secim.groupby('dx')[['dice_baz', 'dice_h3']].agg(['mean', 'std']))
print(secim.groupby('dx')['fark3'].agg(artan=lambda x: (x > 0).sum(),
                                       azalan=lambda x: (x < 0).sum()))
print('Dice<0.05:', secim.groupby('dx')['dice_h3'].apply(lambda x: (x < 0.05).sum()).to_dict())
print(secim.groupby(['dx', 'yedek'])[['dice_baz', 'dice_h3']].agg(['mean', 'count']))
print('Yedek kullanılmayanlarda en büyük |fark|:', secim[~secim['yedek']]['fark3'].abs().max())
```

```python
k = 2
sigma = 0.5 * k

def seg(gri):
    esik, _ = cv2.threshold(gri, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    etiket, n = ndi.label(gri < esik)
    h, w = gri.shape
    if n == 0:
        return np.zeros((h, w), bool)
    secilen = etiket[h // 2, w // 2]
    if secilen == 0:
        secilen = int(np.argmax(np.bincount(etiket.ravel())[1:])) + 1
    return ndi.binary_fill_holes(etiket == secilen)

def buyut(m, h, w):
    return cv2.resize(m.astype(np.uint8), (w, h), interpolation=cv2.INTER_NEAREST) > 0

rows = []
for _, s in secim.iterrows():
    gri = cv2.cvtColor(cv2.imread(s['img']), cv2.COLOR_BGR2GRAY)
    uzman = cv2.imread(s['mask'], cv2.IMREAD_GRAYSCALE) > 127
    h, w = gri.shape
    kw, kh = w // k, h // k
    uk = cv2.resize(uzman.astype(np.uint8), (kw, kh), interpolation=cv2.INTER_NEAREST)
    a = seg(cv2.resize(gri, (kw, kh), interpolation=cv2.INTER_NEAREST))
    b = seg(cv2.resize(cv2.GaussianBlur(gri, (0, 0), sigma), (kw, kh), interpolation=cv2.INTER_AREA))
    c = seg(cv2.GaussianBlur(gri, (0, 0), sigma))
    rows.append(dict(
        dx=s['dx'], genislik=kw, cap=2 * np.sqrt(uk.sum() / np.pi),
        tavan=dice(buyut(uk, h, w), uzman),
        suzgecsiz=dice(buyut(a, h, w), uzman),
        suzgecli=dice(buyut(b, h, w), uzman),
        yalniz=dice(c, uzman),
        bos_a=a.sum() == 0, bos_b=b.sum() == 0, bos_c=c.sum() == 0))

r2 = pd.DataFrame(rows)
r2.to_csv('sonuc_k2.csv', index=False)
dcols = ['tavan', 'suzgecsiz', 'suzgecli', 'yalniz']
print(r2.groupby('dx')[dcols].agg(['mean', 'std']).round(4).to_string())
print(r2.groupby('dx')[['genislik', 'cap']].median())
print(r2.groupby('dx')[['bos_a', 'bos_b', 'bos_c']].sum())
ref = secim.groupby('dx')['dice_h3'].mean()
kayip = (-r2.groupby('dx')[dcols].mean()).add(ref, axis=0)
print((kayip[dcols[1:]]).round(4))
print(kayip[dcols[1:]] > 0.10)
```

```python
def deney(k):
    sigma = 0.5 * k
    rows = []
    for _, s in secim.iterrows():
        gri = cv2.cvtColor(cv2.imread(s['img']), cv2.COLOR_BGR2GRAY)
        uzman = cv2.imread(s['mask'], cv2.IMREAD_GRAYSCALE) > 127
        h, w = gri.shape
        kw, kh = w // k, h // k
        uk = cv2.resize(uzman.astype(np.uint8), (kw, kh), interpolation=cv2.INTER_NEAREST)
        a = seg(cv2.resize(gri, (kw, kh), interpolation=cv2.INTER_NEAREST))
        b = seg(cv2.resize(cv2.GaussianBlur(gri, (0, 0), sigma), (kw, kh), interpolation=cv2.INTER_AREA))
        c = seg(cv2.GaussianBlur(gri, (0, 0), sigma))
        rows.append(dict(
            dx=s['dx'], genislik=kw, cap=2 * np.sqrt(uk.sum() / np.pi),
            tavan=dice(buyut(uk, h, w), uzman),
            suzgecsiz=dice(buyut(a, h, w), uzman),
            suzgecli=dice(buyut(b, h, w), uzman),
            yalniz=dice(c, uzman),
            bos_a=a.sum() == 0, bos_b=b.sum() == 0, bos_c=c.sum() == 0))
    r = pd.DataFrame(rows)
    r.to_csv(f'sonuc_k{k}.csv', index=False)
    dcols = ['tavan', 'suzgecsiz', 'suzgecli', 'yalniz']
    print('k =', k, ' sigma =', sigma)
    print(r.groupby('dx')[dcols].agg(['mean', 'std']).round(4).to_string())
    print(r.groupby('dx')[['genislik', 'cap']].median())
    print(r.groupby('dx')[['bos_a', 'bos_b', 'bos_c']].sum())
    ref = secim.groupby('dx')['dice_h3'].mean()
    kayip = (-r.groupby('dx')[dcols].mean()).add(ref, axis=0)
    print(kayip[dcols[1:]].round(4))
    print(kayip[dcols[1:]] > 0.10)
    return r

r4 = deney(4)
```
